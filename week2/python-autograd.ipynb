```python
import torch

# 创建需要计算梯度的张量
x = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)

# 定义函数：y = x₁² + x₂² + x₃²
y = (x ** 2).sum()

# 反向传播，计算 dy/dx
y.backward()

print("输入张量 x：", x)
print("计算结果 y：", y.item())
print("自动计算的梯度：", x.grad)
print("理论梯度 2x：", 2 * x.detach())

# 验证自动微分结果是否正确
expected_grad = 2 * x.detach()
is_correct = torch.allclose(x.grad, expected_grad)

print("自动微分验证是否通过：", is_correct)
```

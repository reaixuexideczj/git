```python
import torch
import numpy as np

# 1. 加减乘除
a = torch.tensor([1.0, 2.0, 3.0])
b = torch.tensor([4.0, 5.0, 6.0])

print("加法：", a + b)
print("减法：", a - b)
print("逐元素乘法：", a * b)
print("逐元素除法：", a / b)

# 2. 矩阵乘法
A = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
B = torch.tensor([[5.0, 6.0], [7.0, 8.0]])

print("\ntorch.mm：\n", torch.mm(A, B))
print("torch.matmul：\n", torch.matmul(A, B))

# 3. 维度变换
x = torch.arange(12)

print("\n原始形状：", x.shape)
print("view形状：", x.view(3, 4).shape)
print("reshape形状：", x.reshape(2, 6).shape)

y = torch.randn(2, 3, 4)
print("permute前：", y.shape)
print("permute后：", y.permute(1, 0, 2).shape)

# 4. NumPy与张量互相转换
numpy_array = np.array([1, 2, 3], dtype=np.float32)
torch_tensor = torch.from_numpy(numpy_array)
back_to_numpy = torch_tensor.numpy()

print("\nNumPy → Tensor：", torch_tensor)
print("Tensor → NumPy：", back_to_numpy)
```
